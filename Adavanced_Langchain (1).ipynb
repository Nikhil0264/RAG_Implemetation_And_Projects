```python
import os
from google.colab import userdata

from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace

os.environ["HUGGINGFACEHUB_API_TOKEN"] = userdata.get("token")

llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-V4.1-Flash",
    max_new_tokens=300,
    do_sample=False,
    repetition_penalty=1.03,
    provider="auto"
)

chat_model = ChatHuggingFace(llm=llm)
response = chat_model.invoke("What is the capital of India?")

print(response.content)
```

```python
from langchain.tools import tool

@tool
def get_weather(location:str)->str:
  """Get the weather at a location"""
  return f"It's sunny in {location}"

model_with_tools = chat_model.bind_tools([get_weather])
```

```python
response = model_with_tools.invoke("What is the weather in Mumbai?")
print(response.content)
```

```python
print(response)
```

```python
response.tool_calls
```

```python
messages = [{"role":"user","content":"what's the weather in Banglore ?"}]
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)
```

```python
for tool_call in ai_msg.tool_calls:
  tool_result = get_weather.invoke(tool_call)
  messages.append(tool_result)
```

```python
final_response = model_with_tools.invoke(messages)
print(final_response.text)
```

```python
from google.colab import userdata
from langchain_community.utilities import SerpAPIWrapper

search = SerpAPIWrapper(
    serpapi_api_key=userdata.get("searchKey")
)
```

```python
from langchain_core.tools import Tool

search_tool = Tool(
    name="web_search",
    description="Search the web for current or up-to-date information.",
    func=search.run
)
```

```python
@tool("calculator",description="Performs arithemetic calculations")
def calculator(expression:str)->str:
  """Performs arithemetic calculations"""
  return str(eval(expression))
```

```python
from langchain.agents import create_agent
agent = create_agent(
    model = chat_model,
    tools = [search_tool,calculator],
)
```

```python
from pydantic import BaseModel,Field

class Movie(BaseModel):
  title:str=Field(description="The title of the movie")
  year:int=Field(description="The year the movie was released")
  genre:str=Field(description="The genre of the movie")
  director:str=Field(description="The director of the movie")
  rating:float=Field(description="The rating of the movie")
```

```python
structured_model = model.with_structured_output(Movie)
```

```python
structured_model
```

```python
model.invoke("provide detail of movie bahubali")
```

```python
structured_model.invoke("provide detail of movie bahubali")
```

```python
model_with_raw_structure = model.with_structured_output(Movie,include_raw=True)
```

```python
model_with_raw_structure.invoke("provide detail of movie bahubali")
```

```python
class Actor(BaseModel):
  name:str=Field(description="The name of the actor")
  age:int=Field(description="The age of the actor")
  gender:str=Field(description="The gender of the actor")

class MovieDetails(BaseModel):
  title:str=Field(description="The title of the movie")
  year:int
  cast:list[Actor]
  genre:list[str]
  budget:float | None = Field(None,description="the budget in USD")
```

```python
model_with_struc = model.with_structured_output(MovieDetails)
response = model_with_struc.invoke("provide detail of movie bahubali")
print(response)
```

```python
from langchain.agents import create_agent
from langchain.agents.middleware import HumanInTheLoopMiddleware

def read_email_tool(email_id:str)->str:
  """Mock function to read on email by its ID"""
  return f"Email content for ID : {email_id}"

def send_email_tool(recipient:str,subject:str,body:str)->str:
  """Mock function to send mail"""
  return f"Email sent to {recipient} with subject {subject}"
```

```python
agent3 = create_agent(
    model = model,
    tools=[read_email_tool,send_email_tool],
    checkpointer=InMemorySaver(),
    middleware=[
        HumanInTheLoopMiddleware(
            interrupt_on={
                "send_email_tool":{
                    "allowed_decisions":["approve","edit","reject"]
                },
                "read_email_tool":False
            }
        )
    ]

)
```
