```python
import re

with open("video.en.vtt", "r", encoding="utf-8") as f:
    vtt = f.read()

# Remove timestamps, WEBVTT metadata, and formatting
lines = vtt.splitlines()

text_lines = []

for line in lines:
    line = line.strip()

    if not line:
        continue

    if line == "WEBVTT":
        continue

    if "-->" in line:
        continue

    if re.match(r"^\d+$", line):
        continue

    text_lines.append(line)

transcript = " ".join(text_lines)

print(transcript[:5000])
```

```python
with open("/transcript.txt", "r", encoding="utf-8") as f:
    transcript = f.read()

print(transcript)
```

```python
import re

clean_transcript = re.sub(r'\b\d+:\d{2}\b', '', transcript)

# Remove extra spaces/newlines
transcript = re.sub(r'\s+', ' ', clean_transcript).strip()
```

```python
from IPython.display import Markdown
Markdown(transcript)
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunks = splitter.create_documents([transcript])
```

```python
len(chunks)
```

```python
chunks[0]
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
vector_store = FAISS.from_documents(chunks,embeddings)
```

```python
vector_store.index_to_docstore_id
```

```python
vector_store.get_by_ids(['7a1cebcd-a492-4df6-8203-4355df1f384d'])
```

```python
reteriver = vector_store.as_retriever(search_type="similarity", search_kwargs={"k":4})
```

```python
reteriver
```

```python
reteriver.invoke("How to Get Out of Debt?")
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    template="""
You are a helpful transcript assistant.
Answer ONLY from the provided transcript context.
If the context is insufficient, just say you don't know.

Transcript context:
{context}

Question:
{question}

Answer:
""",
    input_variables=["context", "question"]
)
```

```python
question = "is the topic of god discussed in this video?if yes then what was discussed"
retrived_docs = reteriver.invoke(question)
```

```python
retrived_docs
```

```python
context_text = "\n\n".join(doc.page_content for doc in retrived_docs)
```

```python
final_prompt = prompt.invoke({"context":context_text, "question":question})
```

```python
final_prompt
```

```python
result = chat_model.invoke(final_prompt)
```

```python
result.content
```

```python
from langchain_core.runnables import RunnablePassthrough,RunnableParallel,RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

```python
def format_docs(retrived_docs):
  context_text = "\n\n".join(doc.page_content for doc in retrived_docs)
  return context_text
```

```python
parallel_chain = RunnableParallel({
    'context': reteriver | RunnableLambda(format_docs),
    'question':RunnablePassthrough()
})
```

```python
parser = StrOutputParser()
```

```python
main_chain = parallel_chain | prompt | chat_model | parser
```

```python
result = main_chain.invoke(" Can You Manifest Money?")
```
